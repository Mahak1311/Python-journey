## Web Scraping

```python
import requests

URL = "https://www.scrapethissite.com/pages/simple/"

res= requests.get(URL)
```

```python
print(res.status_code) #to get the status code along with response
```

```python
if res.status_code==200:
    print(res.text) #to print html content of the page  
```

```python
if res.status_code==200:
    print(res.content) #to print raw bytes of response
```

```python
if res.status_code==200:
    print(res.headers) #to print extra information
```

```python
with open("scraped-data/data1.html", "w", encoding="utf-8") as f:
    f.write(res.text)
```

## BeautifulSoup

```python

from bs4 import BeautifulSoup

with open("scraped-data/data1.html","r") as f:
    html_content=f.read()

soup = BeautifulSoup(html_content, "lxml")    
```

```python
soup
```
