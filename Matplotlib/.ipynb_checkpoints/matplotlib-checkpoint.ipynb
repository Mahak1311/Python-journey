```python
# Real Data

# Oscar winning movies
Oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232] # in $M

plt.plot(years,oscar_revenue)
plt.title("Oscar Movies Revenue Each Year")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")
```

```python
# Plot multiple
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(years,oscar_revenue,"o--r", label="Oscar Movies") #format strings "marker line colour"
plt.plot(years,non_oscar_revenue,"<-" , color="yellow",linewidth=3,label="Non-Oscar Movies")

plt.title("Oscar Movies Vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")

plt.legend()
```

## Bar Charts(Vertical)

```python
Oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years=[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232] # in $M

plt.bar(years,oscar_revenue, color="pink")
plt.title("Oscar Movies Revenue Each Year")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")

for i in range (len(years)):
    plt.text(years[i],oscar_revenue[i],str(oscar_revenue[i]),ha="center")

plt.ylim(0,max(oscar_revenue)+100)
```

```python
import numpy as np
Oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue=[1005,170,427,133,232] # in $M
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  
years=[2008,2009,2010,2011,2012]
x= np.arange(len(years)) #to arrange numbers evenly
width = 0.4

plt.bar(x-width/2,oscar_revenue,width,label="oscar movies revenue") #shift to LEFT
#we are using np arrays as adding or subtracting value from value is not supported in lists in python
plt.bar(x+width/2,non_oscar_revenue,width,label="non-oscar movie revenues") #shift to RIGHT

plt.title("Oscar vs Non-Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")
plt.legend()
plt.tight_layout()
plt.xticks(x, years) #to replace x with actual data
```

```python
Oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue=[1005,170,427,133,232] # in $M
plt.barh(Oscar_movies,oscar_revenue)
plt.title("oscar movies and their revenues")
plt.xlabel("oscar revenue")
plt.ylabel("oscar movies")
plt.tight_layout()
```

## Scatter Plots

```python
people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
Age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
BP = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x<135 else "red" for x in BP] #list comprehension
plt.scatter(Age,BP,s=BP,alpha=0.9,cmap="OrRd",c=BP) #alpha manages transparency
#plasma is a colormap group 
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label="BP")

for i in range(len(people)):
    plt.annotate(people[i],xy=(age[i],BP[i]),xytext=(age[i]+1,BP[i]+1))
plt.xlim(min(age), max(age)+10)
plt.ylim(min(BP), max(BP)+5)
```

```python
# Multiple scatter plots
cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp,winter_humidity,label="winter")
plt.scatter(summer_temp,summer_humidity,label="summer")
plt.title("summer vs winter")
plt.legend()
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.grid()
```

## Pie charts 

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [5000, 1500, 1200, 1000, 500]

explode = [0, 0, 0, 0.2, 0]

plt.pie(
    amounts,
    labels=expenses,
    autopct="%1.1f%%",
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1,
        "linestyle": "--"
    },
    explode=explode
)

plt.title("Salaries")
plt.tight_layout()
plt.show()
```

```python
# shadow
# plt.pie(amounts, labels=expenses, explode=explode, shadow=True)

# angle
plt.pie(amounts, labels=expenses, explode=explode, shadow=True, startangle=90) # top
plt.pie(amounts, labels=expenses, explode=explode, shadow=True, startangle=180) # flips
```
